## Why Python for ML?
Machine learning code is mostly data loading, transformation, functions, and reusable objects. This notebook reviews those basics.

### Variables and basic data types

A **variable** is a name that refers to a value. Python does not require you to declare a type in advance; it determines the type from the value. A string stores text, an integer stores a whole number, a float stores a decimal number, and a Boolean is either `True` or `False`.

In ML projects, values such as a filename, number of training examples, learning rate, and whether a record passes a rule are stored in variables.

```python
student_name = 'Asha'
study_hours = 3
quiz_score = 82.5
passed = quiz_score >= 70

print(student_name, study_hours, quiz_score, passed)
```

### Lists, loops and conditions

A **list** stores an ordered collection of values. A `for` loop visits each value. An `if/else` condition chooses which code to run based on a Boolean expression.

This pattern is useful for checking multiple records, applying simple rules and iterating over experiment settings. In large datasets, Pandas or NumPy vectorized operations are usually faster and easier to maintain than a Python loop.

```python
scores = [65, 70, 82, 90]
for score in scores:
    if score >= 70:
        print(score, 'pass')
    else:
        print(score, 'review again')
```

### Dictionaries and key-value data

A **dictionary** stores information as key-value pairs. Here, `name`, `hours` and `score` are keys, and their associated values describe one student. Access a value using its key, such as `student['name']`.

Dictionaries are useful for configuration, experiment metadata, mapping category names and collecting model results.

```python
student = {'name': 'Asha', 'hours': 3, 'score': 82}
print(student['name'])
student['passed'] = student['score'] >= 70
student
```

### Functions and return values

A **function** is a named, reusable block of code. Parameters are the inputs it receives; `return` sends a result back to the caller. Returning a value is different from printing it: a returned value can be stored, compared or passed to another function.

This is how we turn data cleaning, feature creation and evaluation steps into reusable operations.

```python
def grade_status(score):
    if score >= 70:
        return 'pass'
    return 'needs practice'

print(grade_status(85))
```

### Reading a CSV with Pandas

A **CSV (comma-separated values)** file is a plain-text table. `pd.read_csv` loads it as a Pandas DataFrame. A **DataFrame** is a table with named columns and indexed rows.

The relative path `datasets/study_log.csv` means the `datasets` folder beside this notebook. The displayed table is an output from the cell; inspect its columns and values before using it in later code.

```python
import pandas as pd

study_log = pd.read_csv('datasets/study_log.csv')
study_log
```

### Classes, objects and methods

A **class** is a blueprint for creating objects. An **object** is one instance created from that blueprint. `__init__` runs when the object is created and stores its initial attributes. `self` refers to the current object. A **method** is a function defined inside a class.

Classes are used in ML libraries to represent estimators and reusable components. For example, `LinearRegression()` creates a model object; calling `fit` trains that object.

```python
class SimpleStudent:
    def __init__(self, name, hours):
        self.name = name
        self.hours = hours

    def add_hours(self, extra_hours):
        self.hours = self.hours + extra_hours

student = SimpleStudent('Asha', 3)
student.add_hours(2)
print(student.name, student.hours)
```

## Key terms

- **Data type:** the kind of value stored, such as text, integer or Boolean.
- **Mutable:** an object that can be changed after it is created; lists and dictionaries are mutable.
- **Function parameter:** a named input a function expects.
- **Return value:** the result a function sends back.
- **DataFrame:** a two-dimensional labelled table from Pandas.
- **Instance:** a particular object created from a class.

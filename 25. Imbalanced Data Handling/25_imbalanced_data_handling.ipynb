# Imbalanced Data Handling

**Objective:** Learn imbalanced data handling with simple, step-by-step examples.

## Imbalanced Data Handling
Imbalanced data means one class is much more common than another. Accuracy can become misleading.

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.model_selection import train_test_split
from sklearn.utils import resample

RANDOM_STATE = 42

data = pd.read_csv('datasets/imbalanced_fraud.csv')
print(data['is_fraud'].value_counts())
```

```python
X = data.drop(columns=['is_fraud'])
y = data['is_fraud']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y)

baseline = LogisticRegression(max_iter=1000)
baseline.fit(X_train, y_train)
print(classification_report(y_test, baseline.predict(X_test)))
```

```python
weighted_model = LogisticRegression(max_iter=1000, class_weight='balanced')
weighted_model.fit(X_train, y_train)
print(classification_report(y_test, weighted_model.predict(X_test)))
```

```python
train_data = X_train.copy()
train_data['is_fraud'] = y_train
majority = train_data[train_data['is_fraud'] == 0]
minority = train_data[train_data['is_fraud'] == 1]

minority_upsampled = resample(
    minority,
    replace=True,
    n_samples=len(majority),
    random_state=RANDOM_STATE
)

balanced_train = pd.concat([majority, minority_upsampled])
print(balanced_train['is_fraud'].value_counts())
```

```python
try:
    from imblearn.over_sampling import SMOTE

    smote = SMOTE(random_state=RANDOM_STATE)
    X_smote, y_smote = smote.fit_resample(X_train, y_train)
    print(pd.Series(y_smote).value_counts())
except ImportError:
    print('Install imbalanced-learn to run SMOTE: pip install imbalanced-learn')
```

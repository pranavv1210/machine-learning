## Why Pandas is useful

Real datasets are usually tables with rows for observations and columns for variables. Pandas provides tools to load these tables, inspect their structure, filter records, calculate summaries, fill missing values and combine related tables.

A **Series** is a one-dimensional labelled sequence. A **DataFrame** is a two-dimensional table made from columns (often Series objects).

## Pandas in ML
Pandas helps you read, inspect, clean, filter, and combine tabular datasets.

### Load a dataset

`read_csv` reads a comma-separated text file into a DataFrame. `data` is the table loaded from the local `students.csv` file. Always inspect a dataset before modeling: a column might contain text, missing values or unexpected categories.

```python
import pandas as pd

data = pd.read_csv('datasets/students.csv')
data
```

### Series

A Series is a single labelled column of values. The `name` argument gives this Series the label `score`. Series objects support vectorized operations, filtering and summary statistics.

```python
scores = pd.Series([78, 88, 69], name='score')
scores
```

### Inspecting a DataFrame

`shape` returns `(number_of_rows, number_of_columns)`. `columns` lists the column labels. `info()` prints column data types and non-null counts. A **null/missing value** represents information that is absent or unknown; missingness should be understood before deciding how to handle it.

```python
print(data.shape)
print(data.columns)
data.info()
```

### Filtering rows

`data['course'] == 'ML'` creates a Boolean mask: one `True` or `False` value for each row. Passing that mask inside `data[...]` selects only rows where the condition is true. This is called **Boolean indexing**.

```python
ml_students = data[data['course'] == 'ML']
ml_students
```

### Grouping and aggregation

`groupby('course')` divides rows into groups that share the same course. `.mean()` calculates the average of the selected numeric columns for each group. This is useful for comparing groups, but always check how many rows contributed to each average.

```python
data.groupby('course')[['hours', 'score']].mean()
```

### Filling missing values

`fillna` replaces missing values. Here, each column's missing values are filled with that column's mean. This is a simple **imputation** strategy. Mean imputation is easy, but it can be distorted by outliers and may reduce apparent variation. For skewed data, median imputation may be more sensible. In a predictive workflow, calculate imputation values on training data only.

```python
clean_data = data.copy()
clean_data['hours'] = clean_data['hours'].fillna(clean_data['hours'].mean())
clean_data['score'] = clean_data['score'].fillna(clean_data['score'].mean())
clean_data
```

### Merging tables

`merge` joins two DataFrames using a shared key, here `course`. This is similar to a database join. The `on` argument names the column that appears in both tables. Check row counts after a merge because duplicate keys can multiply rows.

```python
course_info = pd.DataFrame({
    'course': ['ML', 'Python'],
    'difficulty': ['medium', 'beginner']
})

merged = clean_data.merge(course_info, on='course')
merged
```

## Key terms and checks

- **Series:** one labelled column.
- **DataFrame:** a labelled two-dimensional table.
- **Boolean mask:** a True/False array used to select rows.
- **GroupBy:** split rows into groups and calculate summaries.
- **Imputation:** replacing missing values with an estimated value.
- **Merge/join:** combining tables using matching keys.

Practice: use `.loc` to select rows and columns by labels, `.iloc` to select by integer positions, `value_counts()` to count categories, and `isna().sum()` to count missing values.

# Principal Component Analysis

**Objective:** Learn principal component analysis with simple, step-by-step examples.

## Principal Component Analysis
PCA reduces many numeric columns into fewer columns that keep as much information as possible.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

RANDOM_STATE = 42

data = pd.read_csv('datasets/wine.csv')
X = data.drop(columns=['wine_class'])
y = data['wine_class']
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2, random_state=RANDOM_STATE)
X_pca = pca.fit_transform(X_scaled)

print('Explained variance:', pca.explained_variance_ratio_)
```

```python
pca_data = pd.DataFrame(X_pca, columns=['PC1', 'PC2'])
pca_data['wine_class'] = y
pca_data.head()
```

```python
for wine_class in pca_data['wine_class'].unique():
    subset = pca_data[pca_data['wine_class'] == wine_class]
    plt.scatter(subset['PC1'], subset['PC2'], label=wine_class)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.title('Wine Data After PCA')
plt.show()
```

## Exploratory Data Analysis
EDA helps you understand data before modeling. Look at summaries, distributions, relationships, and unusual values.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv('datasets/wine.csv')
data.head()
```

```python
print(data.shape)
data.describe()
```

```python
data['wine_class'].value_counts()
```

```python
data[['alcohol', 'malic_acid', 'color_intensity']].hist(figsize=(9, 3), bins=15)
plt.tight_layout()
plt.show()
```

```python
sns.boxplot(data=data, x='wine_class', y='alcohol')
plt.title('Alcohol by Wine Class')
plt.show()
```

```python
correlations = data.drop(columns=['wine_class']).corr(numeric_only=True)
sns.heatmap(correlations.iloc[:6, :6], annot=True, cmap='coolwarm')
plt.title('Correlation Sample')
plt.show()
```

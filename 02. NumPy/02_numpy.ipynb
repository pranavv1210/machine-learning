## What NumPy does in a Machine Learning workflow

NumPy is a Python library for efficient numerical arrays. ML data is commonly represented as a matrix: rows are examples and columns are features. A model often performs operations on many numbers at once, so array operations are more efficient and concise than manually looping over every value.

A NumPy `ndarray` is an N-dimensional array. N-dimensional means it can have one, two, three or more axes.

## NumPy in ML
NumPy arrays are fast containers for numbers. Many ML libraries use NumPy internally.

### Creating an array and checking its type

`np.array` converts a Python sequence into an ndarray. `type(scores)` tells you the Python object's class. Unlike a Python list, a NumPy array normally stores one consistent numeric data type, which enables efficient numerical operations.

```python
import numpy as np

scores = np.array([65, 70, 82, 90])
print(scores)
print(type(scores))
```

### Shape, rows, columns and indexing

The `shape` attribute tells how many values exist along each dimension. The matrix in this cell has shape `(2, 3)`: two rows and three columns. NumPy indexing starts at zero, so `matrix[0, 1]` means first row, second column.

The word **axis** means a dimension along which an operation is performed. For a 2-D array, axis 0 refers to rows and axis 1 refers to columns; when aggregating, NumPy reduces the selected axis.

```python
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print(matrix.shape)
print(matrix[0, 1])
```

### Creating ranges and reshaping

`np.arange(12)` creates 12 consecutive values starting at zero. `reshape(3, 4)` changes their layout into three rows and four columns without changing the number of elements. The requested dimensions must multiply to the same element count: 3 × 4 = 12.

Reshaping changes the view of the data, not the underlying values.

```python
values = np.arange(12)
reshaped = values.reshape(3, 4)
reshaped
```

### Broadcasting and element-wise multiplication

The `*` operator performs element-wise multiplication: each row is multiplied by the corresponding weight. NumPy's **broadcasting** rules allow compatible shapes to be combined without manually repeating values. Here, the two weights are applied to each row of the feature matrix.

This is a tiny example of a weighted feature calculation. It is not yet a trained model because the weights were supplied by us.

```python
features = np.array([[10, 2], [20, 4], [30, 6]])
weights = np.array([0.5, 2.0])
features * weights
```

### Aggregations and summary statistics

`mean`, `std` and `max` calculate summary values. The standard deviation (`std`) measures how spread out the values are around their mean. NumPy's default `std` uses population standard deviation (`ddof=0`); Pandas commonly defaults to sample standard deviation. This difference matters when reproducing numerical results.

```python
print('Mean:', scores.mean())
print('Standard deviation:', scores.std())
print('Maximum:', scores.max())
```

### Standardization intuition

The expression subtracts the mean and divides by standard deviation. This is **z-score standardization**: values are expressed in units of standard deviations from the mean. A value near zero is close to average; a positive value is above average; a negative value is below average.

Scaling is important for algorithms that depend on distance or gradient optimization. Fit scaling statistics using training data only to avoid data leakage.

```python
normalized = (scores - scores.mean()) / scores.std()
normalized
```

## Key terms and common mistakes

- **ndarray:** NumPy's N-dimensional array object.
- **Shape:** the number of elements along each dimension.
- **Axis:** the dimension along which a calculation is performed.
- **Broadcasting:** NumPy's rules for combining arrays with compatible shapes.
- **Vectorization:** applying an operation to an array without writing a Python loop.
- **Standardization:** centering values around zero and scaling by standard deviation.

Common mistakes: confusing `*` (element-wise multiplication) with `@` (matrix multiplication), attempting to reshape to an incompatible element count, and standardizing the test set independently rather than reusing training-set statistics.

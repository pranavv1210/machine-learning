# Cross Validation

**Objective:** Learn cross validation with simple, step-by-step examples.

## Cross-validation

A single train/validation split can give a noisy estimate because the result depends on which observations are held out. K-fold cross-validation divides data into K folds. It trains K times, each time holding out a different fold for validation, and summarizes scores.

## Cross Validation
Cross validation trains and tests a model several times on different splits. This gives a more stable estimate than one split.

### Stratified, grouped and time-aware splitting

Stratified K-fold preserves approximate class proportions in each fold and is useful for classification. GroupKFold keeps examples from the same person or group in one fold. TimeSeriesSplit respects chronological order so future observations do not help validate predictions about the past.

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score, StratifiedKFold

RANDOM_STATE = 42

data = pd.read_csv('datasets/iris.csv')
X = data.drop(columns=['species'])
y = data['species']
```

### Choose a scoring metric

The scoring function should reflect the problem. Regression might use negative MAE or negative RMSE in scikit-learn's scoring convention; classification may use accuracy, F1, ROC-AUC or average precision. Scikit-learn often negates losses because its search tools maximize scores.

```python
model = LogisticRegression(max_iter=1000)
folds = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

scores = cross_val_score(model, X, y, cv=folds, scoring='accuracy')
print(scores)
print('Mean accuracy:', scores.mean())
```

## Key terms and cautions

- **Fold:** one partition of the dataset.
- **Validation fold:** the portion held out during one training run.
- **Stratification:** preserving class proportions.
- **Group split:** keeping related records together.
- **Time-aware split:** respecting chronological order.

Put imputation, scaling and feature selection inside a Pipeline so they are fitted only using each fold's training portion.

```python
for fold_number, (train_index, test_index) in enumerate(folds.split(X, y), start=1):
    print('Fold', fold_number, 'train rows:', len(train_index), 'test rows:', len(test_index))
```

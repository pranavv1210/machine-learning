# Train Test Split

**Objective:** Learn train test split with simple, step-by-step examples.

## Train/Test Split
We train on one part of the data and test on unseen data. This gives a more honest estimate of performance.

```python
import pandas as pd
from sklearn.model_selection import train_test_split

RANDOM_STATE = 42

data = pd.read_csv('datasets/iris.csv')
X = data.drop(columns=['species'])
y = data['species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE
)

print('Train rows:', len(X_train))
print('Test rows:', len(X_test))
```

```python
X_train_2, X_test_2, y_train_2, y_test_2 = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE
)

print(X_test.index.equals(X_test_2.index))
```

```python
X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)

print(y.value_counts(normalize=True))
print(y_train_s.value_counts(normalize=True))
print(y_test_s.value_counts(normalize=True))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

model = LogisticRegression(max_iter=1000)
model.fit(X_train_s, y_train_s)
predictions = model.predict(X_test_s)
print('Accuracy:', accuracy_score(y_test_s, predictions))
```

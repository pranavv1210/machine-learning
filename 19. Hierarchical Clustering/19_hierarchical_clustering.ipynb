# Hierarchical Clustering

**Objective:** Learn hierarchical clustering with simple, step-by-step examples.

## Hierarchical Clustering
Hierarchical clustering builds clusters step by step and can be visualized with a dendrogram.

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering

RANDOM_STATE = 42

data = pd.read_csv('datasets/customer_segments.csv')
X = data[['feature_1', 'feature_2']]
data.head()
```

```python
linked = linkage(X, method='ward')
plt.figure(figsize=(10, 4))
dendrogram(linked, truncate_mode='lastp', p=12)
plt.title('Dendrogram')
plt.xlabel('Grouped samples')
plt.ylabel('Distance')
plt.show()
```

```python
model = AgglomerativeClustering(n_clusters=3)
data['predicted_cluster'] = model.fit_predict(X)
data.head()
```

```python
plt.scatter(data['feature_1'], data['feature_2'], c=data['predicted_cluster'])
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Hierarchical Clusters')
plt.show()
```

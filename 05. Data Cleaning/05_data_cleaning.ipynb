**Objective:** Learn data cleaning with simple, step-by-step examples.

## Data cleaning in context

Data cleaning means finding and correcting or documenting issues that could make analysis misleading. Typical issues include missing values, duplicates, inconsistent category names, wrong data types, impossible values and unusual outliers. Do not delete data automatically: first understand why it is unusual.

## Data Cleaning
Real datasets often contain duplicates, missing values, incorrect types, and outliers.

### Load and inspect the messy customer data

Start by viewing the records and checking the dimensions, data types and missing values. A **record** is one row describing an observation. A **schema** describes the expected columns and their types.

```python
import pandas as pd

data = pd.read_csv('datasets/messy_customers.csv')
data
```

### Missing values

A missing value is not the same as zero. It may mean a value was not collected, was not applicable, or was deliberately hidden. The right treatment depends on the reason for missingness and the feature's meaning. Common options are dropping rows, filling with mean/median/mode, using a model-based imputer, or keeping a missingness indicator.

```python
print('Duplicates:', data.duplicated().sum())
data = data.drop_duplicates()
```

### Duplicates and inconsistent text

Duplicate rows can cause the same observation to be counted more than once. Category values can also differ because of spaces, capitalization or spelling. Normalize text carefully, but do not combine categories that are genuinely different.

```python
data['income'] = pd.to_numeric(data['income'], errors='coerce')
print(data.dtypes)
```

### Data types and invalid values

A column that looks numeric may have been loaded as text because one row contains a symbol or word. Convert types only after inspecting the values. Validate domain rules too: an age below zero or a negative quantity may be impossible for the intended dataset.

```python
data['age'] = data['age'].fillna(data['age'].median())
data['income'] = data['income'].fillna(data['income'].median())
data['city'] = data['city'].fillna('Unknown')
data
```

## Key terms and safe workflow

- **MCAR:** missing completely at random; missingness is unrelated to observed or unobserved values.
- **MAR:** missing at random conditional on observed information.
- **MNAR:** missingness depends on the missing value itself or unobserved information.
- **Outlier:** unusually distant observation; it can be an error or a real rare case.
- **Imputation:** replacing missing data with an estimate.

Workflow: inspect → understand why the issue exists → choose a treatment → record the decision → validate the cleaned result. When preparing model data, learn imputation or scaling values on training data only.

```python
income_limit = data['income'].quantile(0.95)
data['income'] = data['income'].clip(upper=income_limit)

data = data[data['age'] < 100]
data
```

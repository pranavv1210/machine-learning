# Matplotlib And Seaborn

**Objective:** Learn matplotlib and seaborn with simple, step-by-step examples.

## Visualization for Learning Data
Charts help you see patterns before choosing a model.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv('datasets/iris.csv')
data.head()
```

```python
species_counts = data['species'].value_counts()
species_counts.plot(kind='bar')
plt.title('Flower Count by Species')
plt.xlabel('Species')
plt.ylabel('Count')
plt.show()
```

```python
data['sepal length (cm)'].plot(kind='hist', bins=15)
plt.title('Sepal Length Distribution')
plt.xlabel('Sepal length')
plt.show()
```

```python
sns.scatterplot(data=data, x='sepal length (cm)', y='petal length (cm)', hue='species')
plt.title('Sepal Length vs Petal Length')
plt.show()
```

```python
correlations = data.drop(columns=['species']).corr()
sns.heatmap(correlations, annot=True, cmap='Blues')
plt.title('Feature Correlation Heatmap')
plt.show()
```
